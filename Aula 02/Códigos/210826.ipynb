```python
import numpy as np
from scipy.fft import fft, fftfreq
import matplotlib.pyplot as plt

def calculate_spectrum(signal, sampling_frequency, single_sided=True):

    N = len(signal) # Number of sample points
    T = 1.0 / sampling_frequency # Sample spacing

    # Perform the FFT
    yf = fft(signal)

    if single_sided:
        xf = fftfreq(N, T)[:N//2] # Frequencies for the positive half of the spectrum
        # Calculate the single-sided amplitude spectrum
        # np.abs(yf[0:N//2]) gets the magnitude of the positive frequencies
        amplitudes = 1.0/N * np.abs(yf[0:N//2])
    else:
        xf = fftfreq(N, T) # All frequencies
        amplitudes = 1.0/N * np.abs(yf)

    return xf, amplitudes

fs = 44100
duration = 1.0
t = np.arange(0, duration, 1/fs)

frequencias = [500, 5000, 10000, 50000]

fig, axs = plt.subplots(4, 1, figsize=(10, 8))
axs = axs.flatten()

for i, f in enumerate(frequencias):

    x = np.cos(2 * np.pi * f * t)

    xf, amplitudes = calculate_spectrum(x, fs, single_sided=True)

    axs[i].plot(xf, amplitudes)
    axs[i].set_title(f"Sinal: {f} Hz")
    axs[i].set_xlabel("Frequência (Hz)")
    axs[i].set_ylabel("Amplitude")
    axs[i].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.signal import chirp
from scipy.fft import fft, fftfreq
import matplotlib.pyplot as plt

def calculate_spectrum(signal, sampling_frequency, single_sided=True):

    N = len(signal) # Number of sample points
    T = 1.0 / sampling_frequency # Sample spacing

    # Perform the FFT
    yf = fft(signal)

    if single_sided:
        xf = fftfreq(N, T)[:N//2] # Frequencies for the positive half of the spectrum
        # Calculate the single-sided amplitude spectrum
        # np.abs(yf[0:N//2]) gets the magnitude of the positive frequencies
        amplitudes = 1.0/N * np.abs(yf[0:N//2])
    else:
        xf = fftfreq(N, T) # All frequencies
        amplitudes = 1.0/N * np.abs(yf)

    return xf, amplitudes

fs = 44100
duration  = .5
t = np.arange(0, duration, 1/fs)
f0 = 500
f1 = 10000

chirp_lin = chirp(t, f0=f0, f1=f1, t1=duration, method='linear')
chirp_quad = chirp(t, f0=f0, f1=f1, t1=duration, method='quadratic')
chirp_log = chirp(t, f0=f0, f1=f1, t1=duration, method='logarithmic')

xf_lin, amp_lin = calculate_spectrum(chirp_lin, fs)
xf_quad, amp_quad = calculate_spectrum(chirp_quad, fs)
xf_log, amp_log = calculate_spectrum(chirp_log, fs)

fig, axs = plt.subplots(3, 1, figsize=(10, 10), sharex=True)

axs[0].plot(xf_lin, amp_lin, color='blue')
axs[0].set_title('Espectro do Chirp: Varredura Linear')
axs[0].set_ylabel('Amplitude')
axs[0].grid(True)

# Plot Quadrático
axs[1].plot(xf_quad, amp_quad, color='green')
axs[1].set_title('Espectro do Chirp: Varredura Quadrática')
axs[1].set_ylabel('Amplitude')
axs[1].grid(True)

# Plot Logarítmico
axs[2].plot(xf_log, amp_log, color='red')
axs[2].set_title('Espectro do Chirp: Varredura Logarítmica')
axs[2].set_xlabel('Frequência (Hz)')
axs[2].set_ylabel('Amplitude')
axs[2].grid(True)

plt.xlim(0, 12000)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from scipy.fft import fft, fftfreq
from IPython.display import Audio, display
import os

def calculate_spectrum(signal, sampling_frequency, single_sided=True):

    N = len(signal)
    T = 1.0 / sampling_frequency

    yf = fft(signal)

    if single_sided:
        xf = fftfreq(N, T)[:N//2]
        amplitudes = 1.0/N * np.abs(yf[0:N//2])
    else:
        xf = fftfreq(N, T)
        amplitudes = 1.0/N * np.abs(yf)

    return xf, amplitudes

def subamostragem(signal, M):

  y = signal[::M]

  return y

M = [2, 4, 8]

sinais_sub = {}
fs_sub = {}

for m in M:
  sinais_sub[m] = subamostragem(signal, m)
  fs_sub[m] = fs // m

fig, axs = plt.subplots(4, 1, figsize=(10, 12), sharex=False)

xf_orig, amp_orig = calculate_spectrum(signal, fs, single_sided=True)
axs[0].plot(xf_orig, amp_orig, color='blue')
axs[0].set_title(f'Sinal Original (fs = {fs} Hz)')
axs[0].set_ylabel('Amplitude')
axs[0].grid(True)

print("Sinal Original:")
display(Audio(signal, rate=fs))

for i, M in enumerate(M):
  ax = axs[i+1]

  xf_sub, amp_sub = calculate_spectrum(sinais_sub[M], fs_sub[M], single_sided=True)


  ax.plot(xf_sub, amp_sub)
  ax.set_title(f'Sinal Subamostrado por M={M} (Nova fs = {fs_sub[M]} Hz)')
  ax.set_ylabel('Amplitude')
  ax.grid(True)

  print(f"\nSinal Subamostrado por M={M}:")
  display(Audio(signal, rate=fs_sub[M]))

  ax.set_xlim(0, fs_sub[M] / 2)

axs[-1].set_xlabel('Frequência (Hz)')
plt.tight_layout()
plt.show()
```

```python
#1

import numpy as np
from scipy.fft import fft, fftfreq
import matplotlib.pyplot as plt

# Funções
##############################################
def calculate_spectrum(signal, sampling_frequency, single_sided=True):

    N = len(signal)
    T = 1.0 / sampling_frequency

    yf = fft(signal)

    if single_sided:
        xf = fftfreq(N, T)[:N//2]
        amplitudes = 1.0/N * np.abs(yf[0:N//2])
    else:
        xf = fftfreq(N, T)
        amplitudes = 1.0/N * np.abs(yf)

    return xf, amplitudes
##############################################

fs = 44100
duration = 1.0
t = np.arange(0, duration, 1/fs)

frequencias = [500, 5000, 10000, 50000]

fig, axs = plt.subplots(4, 1, figsize=(10, 8))
axs = axs.flatten()

for i, f in enumerate(frequencias):

    x = np.cos(2 * np.pi * f * t)

    xf, amplitudes = calculate_spectrum(x, fs, single_sided=True)

    axs[i].plot(xf, amplitudes)
    axs[i].set_title(f"Sinal: {f} Hz")
    axs[i].set_xlabel("Frequência (Hz)")
    axs[i].set_ylabel("Amplitude")
    axs[i].grid(True)

plt.tight_layout()
plt.show()
```

```python
#2

import numpy as np
from scipy.signal import chirp
from scipy.fft import fft, fftfreq
import matplotlib.pyplot as plt

# Funções
##############################################
def calculate_spectrum(signal, sampling_frequency, single_sided=True):

    N = len(signal)
    T = 1.0 / sampling_frequency

    yf = fft(signal)

    if single_sided:
        xf = fftfreq(N, T)[:N//2]
        amplitudes = 1.0/N * np.abs(yf[0:N//2])
    else:
        xf = fftfreq(N, T)
        amplitudes = 1.0/N * np.abs(yf)

    return xf, amplitudes
##############################################

fs = 44100
duration  = .5
t = np.arange(0, duration, 1/fs)
f0 = 500
f1 = 10000

chirp_lin = chirp(t, f0=f0, f1=f1, t1=duration, method='linear')
chirp_quad = chirp(t, f0=f0, f1=f1, t1=duration, method='quadratic')
chirp_log = chirp(t, f0=f0, f1=f1, t1=duration, method='logarithmic')

xf_lin, amp_lin = calculate_spectrum(chirp_lin, fs)
xf_quad, amp_quad = calculate_spectrum(chirp_quad, fs)
xf_log, amp_log = calculate_spectrum(chirp_log, fs)

fig, axs = plt.subplots(3, 1, figsize=(10, 10), sharex=True)

axs[0].plot(xf_lin, amp_lin)
axs[0].set_title('Espectro do Chirp: Varredura Linear')
axs[0].set_ylabel('Amplitude')
axs[0].grid(True)

# Plot Quadrático
axs[1].plot(xf_quad, amp_quad)
axs[1].set_title('Espectro do Chirp: Varredura Quadrática')
axs[1].set_ylabel('Amplitude')
axs[1].grid(True)

# Plot Logarítmico
axs[2].plot(xf_log, amp_log)
axs[2].set_title('Espectro do Chirp: Varredura Logarítmica')
axs[2].set_xlabel('Frequência (Hz)')
axs[2].set_ylabel('Amplitude')
axs[2].grid(True)

plt.xlim(0, 12000)
plt.tight_layout()
plt.show()
```

```python
#8

import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

# Funções
##############################################
def calculate_spectrum(signal, sampling_frequency, single_sided=True):

    N = len(signal)
    T = 1.0 / sampling_frequency

    yf = fft(signal)

    if single_sided:
        xf = fftfreq(N, T)[:N//2]
        amplitudes = 1.0/N * np.abs(yf[0:N//2])
    else:
        xf = fftfreq(N, T)
        amplitudes = 1.0/N * np.abs(yf)

    return xf, amplitudes
##############################################

fsc = 10e6
duration = 0.01
tc = np.arange(0,duration,1/fsc)
xc = np.cos(2000 * np.pi * tc) + np.sin(5000 * np.pi * tc)

f_max = 2500
fs_niq = 2 * f_max
t_niq = np.arange(0,duration,1/fs_niq)
xqui = np.cos(2000 * np.pi * t_niq) + np.sin(5000 * np.pi * t_niq)

x_rec_ret = np.zeros_like(tc)
x_rec_sinc = np.zeros_like(tc)

for i, tn in enumerate(t_niq):

    idx = (tc >= tn) & (tc < tn + 1/fs_niq)
    x_rec_ret[idx] = xqui[i]

for n, xn in enumerate(xqui):

    x_rec_sinc += xn * np.sinc(tc * fs_niq - n)

fig, axs = plt.subplots(4, 1, figsize=(12, 14))

axs[0].plot(tc * 1000, xc, color='blue')
axs[0].set_title('(a) Sinal Contínuo x(t) - Duração de 10 ms')
axs[0].set_xlabel('Tempo (ms)')
axs[0].set_ylabel('Amplitude')
axs[0].grid(True)

xf_cont, amp_cont = calculate_spectrum(xc, fsc)
axs[1].plot(xf_cont, amp_cont, color='purple')
axs[1].set_title('(b) Espectro de x(t) - Aproximação Contínua (10 MHz)')
axs[1].set_xlabel('Frequência (Hz)')
axs[1].set_ylabel('Amplitude')
axs[1].set_xlim(0, 4000)
axs[1].grid(True)

xf_nyq, amp_nyq = calculate_spectrum(xqui, fs_niq)
axs[2].stem(xf_nyq, amp_nyq, basefmt=" ", linefmt='red', markerfmt='ro')
axs[2].set_title('(d) Espectro de x[n] (Amostrado em 5 kHz) - Note o sumiço do pico em 2500 Hz')
axs[2].set_xlabel('Frequência (Hz)')
axs[2].set_ylabel('Amplitude')
axs[2].set_xlim(0, 4000)
axs[2].grid(True)

axs[3].plot(tc * 1000, xc, color='lightgray', label='Original Contínuo', linewidth=3)
axs[3].plot(tc * 1000, x_rec_ret, color='orange', label='Reconstrução Retangular (ZOH)')
axs[3].plot(tc * 1000, x_rec_sinc, color='green', linestyle='dashed', label='Reconstrução Sinc')
axs[3].plot(t_niq * 1000, xqui, 'ko', label='Amostras discretas x[n]')
axs[3].set_title('(e) Sinais Reconstruídos')
axs[3].set_xlabel('Tempo (ms)')
axs[3].set_ylabel('Amplitude')
axs[3].legend(loc='upper right')
axs[3].grid(True)

plt.tight_layout()
plt.show()
```
